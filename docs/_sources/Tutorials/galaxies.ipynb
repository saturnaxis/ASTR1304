## Finding the Distance from a Type Ia Supernova

Type Ia supernovae can shine as bright as an entire galaxy at their maximum brightness, which makes them ideal to use as a standard candle.  **How do astronomers use Type Ia supernovae to estimate distance?**

The first method uses the inverse-square law of light, which relates the flux $F$ (or brightness), luminosity $L$, and distance $d$ by

\begin{align}
F = \frac{L}{4\pi d^2}.
\end{align}

We can do two things to the above formula: (1) rewrite it in terms of the solar constant $F_\odot\ (=1361\ {\rm W/m^2})$ and (2) re-arrange it so that we solve for the distance $d$ directly.  We can get there by,

\begin{align}
\frac{F}{F_\odot} = \left(\frac{L}{L_\odot}\right)\left(\frac{1\ {\rm AU}}{d}\right)^2, \\
\frac{d}{1\ {\rm AU}} = \sqrt{\left(\frac{L}{L_\odot}\right)\left(\frac{1361\ {\rm W/m^2}}{F}\right)}.
\end{align}

From the definition of [parsec](https://en.wikipedia.org/wiki/Parsec), we can use the following conversion factors:

\begin{align*}
1\ {\rm AU} &= \frac{\pi}{6.48 \times 10^5}\ {\rm pc},\\
1\ {\rm AU} &= \frac{\pi}{6.48 \times 10^{11}}\ {\rm Mpc}.
\end{align*}

Type Ia supernovae are typically measured in other galaxies (which are really far away), so let's use the formula with $\rm Mpc$:

\begin{align}
d({\rm Mpc}) = \left(\frac{\pi}{6.48 \times 10^{11}}\right)\sqrt{\left(\frac{L}{L_\odot}\right)\left(\frac{1361\ {\rm W/m^2}}{F}\right)}.
\end{align}

From the above formula, we need to know the luminosity $L$ of a Type Ia supernova and the observed flux $F$ (in $W/m^2$).  The typical maximum luminosity $L$ of a Type Ia supernova is about $9.5 \times 10^9\ L_\odot$.

**In our textbook, it shows a picture of a Type Ia supernova in the Pinwheel Galaxy (M101).  The maximum observed flux of the supernova was $7.5\times 10^{-12}\ {\rm W/m^2}$.  How far away is M101 in ly?**

We can directly substitute the observed flux into our equation and find the distance to M101 in $\rm Mpc$.  Recall that $1\ {\rm pc} = 3.26\ {\rm ly}$, which means that $1\ {\rm Mpc} = 3.26 \times 10^6\ {\rm ly} = 3.26\ {\rm Mly}$.

First, finding the distance in $\rm Mpc$ gives,

\begin{align}
d &= \left(\frac{\pi}{6.48 \times 10^{11}}\right)\sqrt{\left(\frac{9.5 \times 10^9\ L_\odot}{L_\odot}\right)\left(\frac{1361\ {\rm W/m^2}}{7.5\times 10^{-12}\ {\rm W/m^2}}\right)}, \\
&= \left(4.85 \times 10^{-12} \right)\sqrt{\left(9.5 \times 10^9 \right)\left(1.81 \times 10^{14} \right)}, \\
&= \left(4.85 \times 10^{-12} \right)\left(1.31 \times 10^{12} \right), \\
&= 6.4\ {\rm Mpc}.
\end{align}

Now we must convert to $\rm ly$ to get

\begin{align}
d &= \frac{6.4\ {\rm Mpc}}{1}\times \frac{3.26 \times 10^6\ {\rm ly}}{1\ {\rm Mpc}}, \\
&= 2.1 \times 10^7\ {\rm ly}.
\end{align}

Note that M101 is $21\ {\rm Mly}$ away, which means that the supernova occurred about $21\ {\rm million\ years}$ ago.

````python
import numpy as np
L_max = 9.5e9 #maximum luminosity from a Type Ia supernova
F_SN = 7.5e-12 #observed flux from the supernova in M101
F_sun = 1361 #Solar constant (flux from the Sun at 1 AU)

d = (np.pi/6.48e11)*np.sqrt(L_max*F_sun/F_SN)
print("The Type Ia supernova places M101 %1.1f Mpc away." % d)
print("The Type Ia supernova places M101 %1.1e ly away." % (d*3.26e6))
````

## Calculating Recession Velocity and Galactic Distances

We can determine a great deal using the equation for the Doppler effect (see [Making use of the Doppler effect](https://saturnaxis.github.io/IntroAstro/Tutorials/light.html#making-use-of-the-doppler-effect)), where we measure the a spectral feature (in absorption or emission) is shifted in wavelength $\lambda_{\rm obs}$ relative to the rest wavelength $\lambda_{\rm rest}$.  We have the Doppler effect equation for radial velocity $v_r$ as

\begin{align}
v_r = \frac{\lambda_{\rm obs}-\lambda_{\rm rest}}{\lambda_{\rm rest}} \times c,
\end{align}

where $c = 300,000\ {\rm km/s}$, or the speed of light.  Recall that

- $\lambda_{\rm obs}-\lambda_{\rm rest} < 0$ corresponds to a blueshift because $\lambda_{\rm obs}<\lambda_{\rm rest}$, and 
- $\lambda_{\rm obs}-\lambda_{\rm rest}> 0$ corresponds to a redshift because $\lambda_{\rm obs}>\lambda_{\rm rest}$.

For non-relativistic galaxy motion, we can replace the fraction in the Doppler equation with a single value $z$. **Note: there is a correction to $z$ as you approach the speed of light.**  As a result, we now have

\begin{align}
v_r = z \times c.
\end{align}

**You're observing nebulae in the 1920's and taking measurements of using a spectrograph (see {numref}`galaxy_fig` below).**  You try to measure the $H\ \alpha$ line at $656.3\ {\rm nm}$ but find that it's redshifted (see {numref}`tab-radial-velocity` below).

Let's use the data for NGC 5236, where $\lambda_{\rm obs} = 657.394\ {\rm nm}$ and $\lambda_{\rm rest} = 656.3\ {\rm nm}$.  From those values, we can determine $z$ by

\begin{align*}
z = \frac{657.394 - 656.3}{656.3} = \frac{1.094}{656.3} = 0.0017.
\end{align*}

Next, we can find the radial velocity by

$$ v_r = 0.0017 \times (3\times 10^5\ {\rm km/s}) = 500\ {\rm km/s},$$

which matches our table.  Look's like we're doing it right.

Now, let's find the distance to NGC 5236 using the Hubble constant $(H_o = 70\ {\rm km/s/Mpc})$.  Hubble's law is given by

\begin{align}
v_r = H_o \times d,
\end{align}

where $d$ is the distance in $Mpc$.  Using a little algebra, we find for NGC 5236

$$ d = \frac{500\ {\rm km/s}}{70\ {\rm km/s/Mpc}} = 7.1\ {\rm Mpc} = 7.1 \times 10^6\ {\rm pc}.$$

From the measurement of a spectral line, we can find that NGC 5236 is $7.1\ {\rm Mpc}$ away.  For comparision, Andromeda Galaxy (M31) is only $0.77\ {\rm Mpc}$ away, which means that NGC 5236 is about $10\times$ more distant than M31.


## Size & Density of a SMBH
### Size of a SMBH
The size of a black hole depends only on a single parameter, its **mass**.  Recall the expression for the Schwarzschild radius $R_s$, which relates the radius of the black hole (or event horizon) to the black hole mass $M_{\rm BH}$, as

\begin{align}
R_s = \frac{2 G M_{\rm BH}}{c^2} \approx 3\ {\rm km} \times \left(\frac{M_{\rm BH}}{M_\odot}\right),
\end{align}

where $G$ represents the [contant of universal gravitation](https://en.wikipedia.org/wiki/Gravitational_constant) and $c$ is the speed of light.

The largest supermassive black holes (SMBHs) observed have about $10\ {\rm billion}\ M_\odot$.  For example, the black hole at the center of [M87](https://en.wikipedia.org/wiki/Messier_87) (a supergiant elliptical galaxy in Virgo) has a mass of $6.69 \times 10^9 M_\odot$.  We can easily find the size of this SMBH as

$$ R_s = 3\ {\rm km} \times \left(\frac{6.69 \times 10^9 M_\odot}{M_\odot}\right) = 2.0 \times 10^{10}\ {\rm km}, $$

or $20\ {\rm billion km}$.  We can convert that value in to astronomical units (AU), where $1\ {\rm AU} \approx 1.5\times 10^{8}\ {\rm km}$, by

$$R_s = \frac{2 \times 10^{10}\ {\rm km}}{1} \times \frac{1\ {\rm AU}}{1.5\times 10^8\ {\rm km}} = 130\ {\rm AU}. $$

The radius of the M87's SMBH is large enough to engulf all of the Solar System planets (recall $a_{\rm Nept} \approx 30\ {\rm AU}$).  It's big enough to engulf a sizeable portion of the Kuiper belt too.

```{note}
We know that light takes $8.3\ {\rm min}$ to travel $1\ {\rm AU}$ (or the distance from the Earth to the Sun).  This means that it would take light about $1080$ light-minutes for light to traverse just the radius of M87, or around $18\ {\rm hours}$ to travel from the event horizon to the center of the SBMH.
```

### Density of a SMBH
How compact (or dense) is the mass inside the event horizon?  The mass of a SMBH can be measured by measuring the orbital period and semimajor axis of the star's closest to them.  The bulk (average) density of an object is determined by spreading all the mass uniformly into a sphere of radius $R$.  In this case, the radius of the sphere is the Schwarzschild radius $R_s$, so that the density $\rho_{\rm BH}$ is given by,

\begin{align}
\rho_{\rm BH} &= \frac{M_{\rm BH}}{\frac{4}{3}\pi R_s^3},\ \text{or} \\
&= \left(\frac{M_{\rm BH}}{M_\odot}\right) \left(\frac{R_s}{R_\odot}\right)^{-3} \rho_\odot,
\end{align}

where we can make an approximation of the Schwarzschild radius $R_s$ and make a substitution to get,

\begin{align}
\rho_{\rm BH} &\approx \left(\frac{R_\odot}{3\ {\rm km}}\right)^3 \left(\frac{M_{\rm BH}}{M_\odot}\right)^{-2}\rho_\odot, \\
&= 1.25 \times 10^{16} \left(\frac{M_{\rm BH}}{M_\odot}\right)^{-2}\rho_\odot,
\end{align}

where $R_\odot$ is a [solar radius](https://sco.wikipedia.org/wiki/Solar_radius) ($6.957 \times 10^5\ {\rm km}$) and $\rho_\odot$ is the [solar density](http://hyperphysics.phy-astr.gsu.edu/hbase/Solar/sun.html) $(1.41\ {\rm g/cm^3})$.

**Consider the density of different SMBHs ranging from the Milky Way (${\sim}4 \times 10^6\ {\rm M_\odot}$) to M87.**  You want to express the density of the black hole $\rho_{\rm BH}$ in normal units (i.e., $\rm g/cm^3$) and astronomical units $(\text{i.e., } M_\odot/{\rm AU}^3)$.  Note that the mass of the M87's SMBH is about $1673\times$ the mass of the Milky Way's SMBH.

Let's find the density of the SMBH at the center of the Milky Way $\rho_{\rm MW}$.  Using our formula from above,

\begin{align*}
\rho_{\rm MW} &= 1.25 \times 10^{16} \left(\frac{4\times 10^6 M_\odot}{M_\odot}\right)^{-2} \rho_\odot, \\
&= 781\rho_\odot = 1100\ {\rm g/cm^3} = 1.1 \times 10^6\ {\rm kg/m^3}.
\end{align*}

The SMBH of our own Milky Way is about $781\times$ denser than the Sun.

Now let's consider a more massive SMBH like the one at the center of M87.  Since, we know about how much more massive the SMBH is compared to the Milky Way, we can use make use of our previous result by

\begin{align*}
\rho_{\rm M87} &= \left(\frac{\rm M_{\rm M87}}{\rm M_{\rm MW}}\right)^{-2}\rho_{\rm MW},\\
&= \left(\frac{1}{1673} \right)^2 \left(781 \rho_\odot \right), \\
&= 2.79 \times 10^{-4} \rho_\odot = 3.9 \times 10^{-4}\ {\rm g/cm^3} = 0.39\ {\rm kg/m^3}.
\end{align*}

That is about $1/3$ the density of air at sea level $(\rho_{\rm air} = 1225\ {\rm kg/m^3})$ or about $1/2500$ the density of water.  Lower mass black holes are denser than SMBHs because the density scales with the inverse square of the mass, $\rho_{\rm BH} \propto 1/M_{\rm BH}^2$.

```{note}
The conversion to astronomical units is straightfoward where you should find that $1\ {\rm g/cm^3} = 1.68 \times 10^6\ M_\odot/{\rm AU}^3.$  In these units the solar density $\rho_\odot = 2.36\ M_\odot/{\rm AU}^3.$  This seems like a lot, but remember that the Sun's radius $R_\odot = 0.00465\ {\rm AU}$ is really small.
```
